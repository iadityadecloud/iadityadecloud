```python
#PySpark code to find duplicate records with dummy data

from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window

spark = SparkSession.builder.getOrCreate()

departments_data = [
    (1, "HR"),
    (2, "IT"),
    (3, "Finance"),
    (4, "Sales")
]

departments_df = spark.createDataFrame(
    departments_data,
    ["dept_id", "dept_name"]
)

departments_df.show()

employees_data = [
    (101, "Amit", 90000, 2, None, "2020-01-10"),
    (102, "Neha", 80000, 2, 101, "2021-03-15"),
    (103, "Ravi", 75000, 2, 101, "2022-07-20"),
    (104, "Priya", 70000, 1, None, "2019-11-01"),
    (105, "Kiran", 50000, 1, 104, "2022-02-10"),
    (106, "Arjun", 60000, 3, None, "2020-05-18"),
    (107, "Meena", 55000, 3, 106, "2023-01-25"),
    (108, "Suresh", 65000, 4, None, "2021-06-30"),
    (109, "Divya", 48000, 4, 108, "2023-08-12"),
    (110, "Rahul", 47000, 4, 108, "2024-02-05")
]


#--PySpark code to find duplicate records with dummy data

employee_df =spark.createDataFrame(
        employees_data, ["emp_id","emp_name","salary","dept_id","manager_id","join_date"]).withColumn("join_date",F.to_date("join_date"))

# withColumn("join_date",F.to_date("join_date") -- converts a string column into a proper date datatype.

employee_df.show()

windows_spec = Window.orderBy(F.col("salary").desc())

result_df =( employee_df.withColumn("rank",F.dense_rank().over(windows_spec))
        .filter(F.col("rank")==2)
        .select("emp_name","salary")
)                    
result_df.show()

```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,count
spark =SparkSession.builder.appName("FindDuplicates").getOrCreate()

data = [(1,"Ravi", "IT", 50000),(2,"Anu", "HR",60000),(3,"Ravi","IT",50000),(4,"Meera","Finance",70000)]

columns = ["emp_id","emp_name","department","salary"]

df =spark.createDataFrame(data,columns)

df.show()

#Find duplicates records based on emp_name,department, salary

duplicates = ( 
              df.groupby("emp_name","department","salary")
              .count()
              .filter(col("count")>1)

)

duplicates.show()

# To display full duplicate rows
duplicate_rows = (
                df.join(duplicates.select("emp_name","department","salary"),
                        on =["emp_name", "department", "salary"],
                        how="inner"
                        )
)
duplicate_rows.show()
```
